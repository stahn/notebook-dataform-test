```python
# prompt: append to table `lexmotion-analytics.test_repo.notebook_output` a row with the current timestamp and a random string. the schema of the table is as follow:
# Field name
# Type
# Mode
# Key
# Collation
# Default Value
# Policy Tags
# Description
# time
# DATETIME	NULLABLE	-	-	-
# -
# -
# output
# STRING	NULLABLE	-	-	-
# -
# -

import pandas_gbq
import pandas as pd
from datetime import datetime
import random
import string

# Generate random string
random_string = ''.join(random.choices(string.ascii_lowercase + string.digits, k=10))

# Create a DataFrame with the new data
df_new = pd.DataFrame({'time': [datetime.now()], 'output': [random_string]})

# Append the DataFrame to the BigQuery table
pandas_gbq.to_gbq(df_new, 'lexmotion-analytics.test_repo.notebook_output', project_id='lexmotion-analytics', if_exists='append')
```
